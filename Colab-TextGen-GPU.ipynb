```python
import os
import urllib.request
import zipfile

# 1. Limpa resíduos anteriores para evitar conflitos de arquivos
if os.path.exists("main.zip"): os.remove("main.zip")

print("--> Baixando os arquivos oficiais do painel TextGen...")
# URL atualizada e corrigida pós-renomeação do projeto
url_painel = "https://github.com/oobabooga/text-generation-webui/archive/refs/heads/main.zip"

# Configura o cabeçalho para o GitHub aceitar a requisição do Python sem bloquear
req = urllib.request.Request(url_painel, headers={'User-Agent': 'Mozilla/5.0'})
with urllib.request.urlopen(req) as response, open("main.zip", "wb") as out_file:
    out_file.write(response.read())

# 2. Descompactando o arquivo ZIP real
print("--> Extraindo arquivos do servidor...")
with zipfile.ZipFile("main.zip", 'r') as zip_ref:
    zip_ref.extractall(".")

# 3. Organizando as pastas estruturais
print("--> Configurando diretórios do sistema...")
if os.path.exists("textgen"):
    !rm -rf textgen
os.rename("textgen-main", "textgen") # Corrected from 'text-generation-webui-main'
os.chdir("textgen")

# 4. Criando a pasta interna para armazenar os modelos de IA
os.makedirs("user_data/models", exist_ok=True)

# 5. Baixando o modelo Qwen 3.5 (Excelente, leve e sem restrições de fábrica)
print("--> Baixando o cérebro da Inteligência Artificial (Aguarde de 1 a 3 minutos)...")
model_url = "https://huggingface.co/unsloth/Qwen3.5-9B-GGUF/resolve/main/Qwen3.5-9B-Q4_K_M.gguf"
req_model = urllib.request.Request(model_url, headers={'User-Agent': 'Mozilla/5.0'})
with urllib.request.urlopen(req_model) as response, open("user_data/models/Qwen3.5-9B-Q4_K_M.gguf", "wb") as out_file:
    out_file.write(response.read())

# 6. Rodando o instalador e gerando a sua interface web de chat
print("--> Instalando dependências na GPU e ligando o servidor (Aguarde alguns minutos)...")
!chmod +x start_linux.sh
!./start_linux.sh --share --model Qwen3.5-9B-Q4_K_M.gguf

```
